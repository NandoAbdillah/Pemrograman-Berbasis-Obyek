```python
class Mahasiswa :
    def __init__(self, nama, umur, nim, jurusan):
        self.nama = nama
        self.umur = umur
        self.nim = nim
        self.jurusan = jurusan
        print("Mahasiswa berhasil dibuat")
        
    def perkenalan(self):
        print(f"Nama saya {self.nama}, NIM saya {self.nim}, dan saya jurusan {self.jurusan}")
        
    def set_nama(self, namBaru):
        self.nama = namBaru
        
    def set_umur(self, umurBaru) :
        self.umur = umurBaru
        
    def set_nim(self, nimBaru):
        self.nim = nimBaru
        
    def set_jurusan(self, jurusanBaru) :
        self.jurusan = jurusanBaru
        
    def get_nama(self):
        return self.nama
    
    def get_umur(self):
        return self.umur
    
    def get_nim(self):
        return self.nim
    
    def get_jurusan(self):
        return self.jurusan
    
```

```python
# CONSTRUCTOR
mhs1 = Mahasiswa("Nando", 20, "25051204012", "Teknik Informatika")
mhs1.perkenalan()
```

```python
# SETTER
mhs1.set_nama("Albani Rassyeah")
mhs1.set_umur(21)
mhs1.set_nim("25051204013")
mhs1.set_jurusan("Sastra Mesin")
```

```python
# GETTER 
print(f"Nama Mahasiswa: {mhs1.get_nama()}")
print(f"Umur Mahasiswa: {mhs1.get_umur()}")
print(f"NIM Mahasiswa: {mhs1.get_nim()}")
print(f"Jurusan Mahasiswa: {mhs1.get_jurusan()}")
```
