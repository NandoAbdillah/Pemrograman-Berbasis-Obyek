```python
class Mahasiswa :
    def __init__(self, nama, umur, nim, jurusan):
        self.__nama = nama
        self.__umur = umur
        self.__nim = nim
        self.__jurusan = jurusan
        print("Mahasiswa berhasil dibuat")
        
    def perkenalan(self):
        print(f"Nama saya {self.__nama}, NIM saya {self.__nim}, dan saya jurusan {self.__jurusan}")
        
    def set_nama(self, namBaru):
        self.__nama = namBaru
        
    def set_umur(self, umurBaru) :
        self.__umur = umurBaru
        
    def set_nim(self, nimBaru):
        self.__nim = nimBaru
        
    def set_jurusan(self, jurusanBaru) :
        self.__jurusan = jurusanBaru
        
    def get_nama(self):
        return self.__nama
    
    def get_umur(self):
        return self.__umur
    
    def get_nim(self):
        return self.__nim
    
    def get_jurusan(self):
        return self.__jurusan
    
    
```

```python
mhs1 = Mahasiswa("Budi", 20, "123456789", "Informatika")
print(mhs1.get_nama())
```
