# Inheritance

```python
class Mahasiswa :
    def __init__(self, nama, umur, nim, jurusan):
        self.nama = nama
        self.umur = umur
        self.nim = nim
        self.jurusan = jurusan
        print("Mahasiswa berhasil dibuat")
        
    def perkenalan(self):
        print(f"Nama saya {self.nama}, NIM saya {self.nim}, dan saya jurusan {self.jurusan}")
        
    def set_nama(self, namBaru):
        self.nama = namBaru
        
    def set_umur(self, umurBaru) :
        self.umur = umurBaru
        
    def set_nim(self, nimBaru):
        self.nim = nimBaru
        
    def set_jurusan(self, jurusanBaru) :
        self.jurusan = jurusanBaru
        
    def get_nama(self):
        return self.nama
    
    def get_umur(self):
        return self.umur
    
    def get_nim(self):
        return self.nim
    
    def get_jurusan(self):
        return self.jurusan
    
```

```python
mhs1 = Mahasiswa("Andi", 20, "123456789", "Informatika")
```

```python
mhs1.perkenalan()
```

```python
class MahasiswaInternasional(Mahasiswa) :
    def __init__(self, nama, umur, nim, jurusan, negera_asal):
        super().__init__(nama, umur, nim, jurusan)
        self.negera_asal = negera_asal

    #OVERRIDE
    def perkenalan(self):
        print(f"Nama saya {self.nama}, NIM saya {self.nim}, saya jurusan {self.jurusan}, dan saya berasal dari {self.negera_asal}")

    def perkenalan(self, mata_uang) :
        print(f"Nama saya {self.nama}, NIM saya {self.nim}, saya jurusan {self.jurusan}, dan saya berasal dari {self.negera_asal}. Mata uang negara saya adalah {mata_uang}")
```

```python
mhs2 = MahasiswaInternasional("Budi", 22, "987654321", "Sistem Informasi", "Kamboja")
```

```python
mhs2.perkenalan("Riel")
```
