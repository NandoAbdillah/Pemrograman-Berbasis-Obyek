```python
class GameObject :
    def __init__(self, symbol, row, col):
        self.symbol = symbol
        self.row = row
        self.col = col
        
    def print_object(self):
        print(self.symbol, self.row, self.col)
        
```

```python
class Grid :
    def __init__ (self, rows, cols) :
        self.rows = rows
        self.cols = cols
        self.board = []
        
        for i in range(rows) :
            row = []
            for j in range(cols) :
                row.append("-")
            self.board.append(row)
            
    def display(self) :
        for row in self.board :
            print(" ".join(row))
            
    def place_object(self, obj) :
        self.board[obj.row][obj.col] = obj.symbol
    
    def move_object(self, obj, nrow, ncol) :
        if not self.is_position_valid( nrow, ncol) :
            print( f"Posisi baru tidak valid melibihi atau kurang dari ukuran grid {self.rows} x {self.cols} !" )
            return
            
        self.board[obj.row][obj.col] = "-"
        
        obj.row = nrow
        obj.col = ncol
        
        self.board[obj.row][obj.col] = obj.symbol
        
    def is_position_valid(self, nrow, ncol) :
        return 0 <= nrow < self.rows and 0 <= ncol <self.cols
```

```python
grid_row = 10
grid_col = 10

grid = Grid(grid_row, grid_col)
Grid.display(grid)
```

```python
#  Meletakkan Objek Miya

miya = GameObject("M", 5, 2)

Grid.place_object(grid, miya)
Grid.display(grid)
```

```python
# Memindahkan Miya ke posisi baru yang valid

nrow = 5
ncol = 2

Grid.move_object(grid, miya, nrow, ncol)
Grid.display(grid)
```

```python
# Memindahkan Miya ke posisi baru yang tidak valid

nrow = 15
ncol = 20

Grid.move_object(grid, miya, nrow, ncol)
```
