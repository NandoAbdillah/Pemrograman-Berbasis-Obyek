## Day 2 Python Objek

```python
class Mahasiswa :
    def __init__(self, nama, nim, kelas, prodi, umur):
        self.nama = nama
        self.nim = nim
        self.kelas = kelas
        self.prodi = prodi
        self.umur = umur

    def memperkenalkan(self) :
        print("Nama saya adalah " , self.nama, 
              "\nUmur saya adalah " , self.umur,
              "\nNIM saya adalah " , self.nim,
              "\nKelas saya adalah " , self.kelas,
              "\nProdi saya adalah " , self.prodi)
        
    def umur_2031(self):
        print(self.nama , "pada tahun 2031 berumur", (2031-2026) + self.umur, "tahun")
```

```python
mhs1 = Mahasiswa("Nando Abdillah Salam", "25051204012", "2025G", "Teknik Informatika", 20)
```

```python
mhs1.memperkenalkan()
```

```python
mhs2 = Mahasiswa("Yusuf", "25051204012", "2024A", "Sastra Mesin", 28)
mhs2.memperkenalkan()
```

```python
mhs1.umur_2031()
mhs2.umur_2031()
```
